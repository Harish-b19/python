```python
from functools import reduce
```

```python
#1. Using reduce(), calculate the sum of all numbers.
def sum(n,n1):
  return n+n1
n=[10, 20, 30, 40, 50]
r=reduce(sum,n)
print(r)
```

```python
#2. Using reduce(), calculate the product of all numbers.
def prod(n2,n3):
  return n2*n3
n2= [2, 3, 4, 5]
r1=reduce(prod,n2)
print(r1)
```

```python
#3. Using reduce(), find the maximum number.
def ma(x,y):
  if x>y:
    return x
  else:
    return y
n4= [45, 12, 89, 34, 67, 91, 28]
r2=reduce(ma,n4)
print(r2)
```

```python
#4. Using reduce(), find the minimum number.
def mi(x1,y1):
  if x1<y1:
    return x1
  else:
    return y1
n5=[45, 12, 89, 34, 67, 91, 28]
r3=reduce(mi,n5)
print(r3)
```

```python
#5. Using reduce(), combine all strings into one sentence.
def con(x2,y2):
  return x2+y2
n6= ["Python", " ", "is", " ", "powerful"]
r4=reduce(con,n6)
print(r4)
```

```python
#6. Using reduce(), calculate the total number of characters in all words.
def cal(w,y):
  return w+len(y)
n7= ["Python", "Azure", "Cloud", "AI"]
r5=reduce(cal,n7,0)
print(r5)
```

```python
#7. Using reduce(), calculate the factorial of number.
def fact(a,b):
  return a*b
#n8=6
n8=list(range(1,6+1))
r6=reduce(fact,n8)
print(r6)
```

```python
#8. Using reduce(), calculate the sum of only the even numbers. Do not create a separate filtered list.
def even(a5,b5):
  if b5%2==0:
    return a5 + b5
  else:
    return a5

n9=[12, 45, 67, 89, 23, 56, 91, 34, 78, 100]
r7=reduce(even,n9,0)
print(r7)
```

```python
#9. Using reduce(), calculate the sum of only the odd numbers. Do not create a separate filtered list.
def odd (a6,b6):
  if b6%2:
    return a6+b6
  else:
    return a6
n10= [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]
r8=reduce(odd,n10,0)
print(r8)
```

```python
#10. Using reduce(), find the longest word
def lon(a9,b9):
  if(len(b9)>len(a9)):
    return b9
  else:
    return a9
n11= ["AI", "Python", "Databricks", "Cloud", "Kubernetes"]
r9=reduce(lon,n11)
print(r9)
```

```python
#11. Using reduce(), find the shortest word.
def short(a10,b10):
  if len(a10)<len(b10):
    return a10
  else:
    return b10
n12=   ["AI", "Python", "Databricks", "Cloud", "Kubernetes"]
r10=reduce(short,n12)
print(r10)
```

```python
#14. Using reduce(), find the employee with the highest salary
def high(v2,t2):
  if v2>t2["salary"]:
    return v2
  else:
    return t2["salary"]
n15= [
    {"name": "Rahul", "salary": 45000},
    {"name": "Anil", "salary": 65000},
    {"name": "Sneha", "salary": 72000},
    {"name": "Kiran", "salary": 82000},
    {"name": "Meena", "salary": 55000}
]
r13=reduce(high,n15,0)
print(r13)
```

```python
#17. Using reduce(), calculate the total inventory value using price × stock.
def inven(v5,t5):
  return v5+(t5["stock"]*t5["price"])
n18=[
    {"name": "Laptop", "price": 55000, "stock": 5},
    {"name": "Mouse", "price": 600, "stock": 25},
    {"name": "Keyboard", "price": 1500, "stock": 15},
    {"name": "Monitor", "price": 18000, "stock": 8}
]
r16=reduce(inven,n18,0)
print(r16)
```

```python
#19.Using reduce(), merge all dictionaries into one dictionary.
def merge(v7,t7):
  return v7|t7
n20= [
    {"name": "Rahul"},
    {"age": 30},
    {"department": "IT"},
    {"salary": 65000}
]
r17=reduce(merge,n20)
print(r17)
```

```python
from functools import reduce
```

```python
#1. Using filter() and map():

#Extract all even numbers.
#Calculate the square of every even number.
#Return the squared values as a list.

def even(n):
  if n%2==0:
    return n
def sq(n):
  return n**2
n=[12, 45, 67, 89, 23, 56, 91, 34, 78, 100]
r=list(map(sq,filter(even,n)))
print(r)
```

```python
#2. Using filter() and reduce():

#Extract all odd numbers.
#Calculate their total sum.
def odd(n1):
  if n1%2:
    return n1
def a(x,y):
  return x+y
n1=[12, 45, 67, 89, 23, 56, 91, 34, 78, 100]
r1=reduce(a,list(filter(odd,n1)))
print(r1)
```

```python
'''#3. Using filter() and map():
     Extract numbers greater than 50.
     Multiply every selected number by 2 '''
def great(n2):
  if n2>50:
    return n2
def sel(n2):
  return n2*2
n2=[10, 25, 40, 55, 70, 85, 100]
r2=list(map(sel,filter(great,n2)))
print(r2)
```

```python
'''#4. using filter(), map(), and reduce():

Extract positive numbers.
Calculate the square of every positive number.
Calculate the total sum of the squared values. '''
def pos(n3):
  return n3>0
def sq(n3):
  return n3**2
def s(x,y):
  return x+y
n3=[-15, 20, -35, 40, 50, -10, 60]
r3=reduce(s,list((map(sq,list(filter(pos,n3))))))
print(r3)
```

```python
'''#5.Using filter() and map():

Extract words containing more than five characters.
Convert the selected words to uppercase.'''
def ch(n4):
  return len(n4)>5
def w(n4):
  return n4.upper()
n4=["python", "azure", "ai", "developer", "cloud", "analytics", "spark"]
r4=list(map(w,list(filter(ch,n4))))
print(r4)
```

```python
'''#6. Using filter() and map():

Extract names beginning with "A".
Convert the selected names to lowercase.'''
def beg(n5):
  return n5.startswith('A')
def low(n5):
  return n5.lower()
n5=["Anil", "Rahul", "Ajay", "Meena", "Arjun", "Kiran", "Akhil"]
r5=list(map(low,list(filter(beg,n5))))
print(r5)
```

```python
'''#7. Using filter() and reduce():

Extract numbers divisible by both 3 and 5.
Calculate their total sum.'''
def div(n6):
  return ((n6%3==0)and (n6%5==0))

def su(x,y):
  return x+y
n6=  [3, 5, 10, 15, 18, 30, 42, 45, 60, 75]
r6=reduce(su,list(filter(div,n6)))
print(r6)
```

```python
'''#8. Using filter() and reduce():

Extract palindrome words.
Join them into one string separated by " | ".'''
def pal(n7):
  return n7==n7[::-1]
def j(x,y):
  return x+"|"+y
n7= ["madam", "python", "level", "azure", "radar", "cloud", "civic"]
r7=reduce(j,list(filter(pal,n7)))
print(r7)
```

```python
'''#9. Using filter(), map(), and reduce():

Extract temperatures greater than 30°C.
Convert the selected values into Fahrenheit.
Calculate the average Fahrenheit temperature.'''
def ce(n8):
  return n8>30
def fah(n8):
  return (n8*9/5)+32
def avg(x,y):
  return (x+y)
n8=[18, 24, 30, 35, 40, 12, 28, 33, 15, 42]
k=list(map(fah,list(filter(ce,n8))))
r7=reduce(avg,k,0)/len(k)
print(r7)
```

```python
'''#10. Using filter(), map(), and reduce():

Extract prices greater than 1000.
Apply a 10% discount to every selected price.
Calculate the total discounted amount.'''
def g(n9):
  return n9>1000
def dis(n9):
  return n9-(n9*0.1)
def am(x1,y1):
  return x1+y1

n9=[450, 1200, 2500, 800, 5000, 1500, 7500]
r9=reduce(am,list(map(dis,list(filter(g,n9)))),0)
print(r9)
```

```python
''' #16
Using filter(), map(), and reduce():

Extract products currently in stock.
Calculate each product's inventory value using price × stock.
Calculate the total inventory value.'''

def cur(n15):
  return n15["stock"]>0
def inven(n15):
  return n15["price"]*n15["stock"]
def val(a5,b5):
  return a5+b5

n15=[
    {"name": "Laptop", "price": 55000, "stock": 5},
    {"name": "Mouse", "price": 600, "stock": 25},
    {"name": "Keyboard", "price": 1500, "stock": 0},
    {"name": "Monitor", "price": 18000, "stock": 8},
    {"name": "Printer", "price": 12000, "stock": 0},
    {"name": "SSD", "price": 6500, "stock": 12}
]
r16=reduce(val,list(map(inven,list(filter(cur,n15)))))
print(r16)
```

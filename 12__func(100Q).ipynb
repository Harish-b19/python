```python
'''Assignment 92: Closure-Based Bank Balance
Write a function named create_bank_account().

Requirements
Store the balance in the outer function.
Define inner functions for:
Deposit
Withdrawal
Balance enquiry
Use nonlocal to update the balance.
Reject withdrawals greater than the available balance.
Return all three inner functions.'''
def create_bank_account():
    balance = 0

    def deposit(amount):
        nonlocal balance
        balance += amount
        print("Deposited:", amount)

    def withdrawal(amount):
        nonlocal balance
        if amount > balance:
            print("Insufficient balance")
        else:
            balance -= amount
            print("Withdrawn:", amount)

    def balance_enquiry():
        return balance

    return deposit, withdrawal, balance_enquiry


deposit, withdrawal, balance = create_bank_account()

deposit(1000)
print("Balance:", balance())

withdrawal(300)
print("Balance:", balance())

withdrawal(800)

deposit(500)
print("Balance:", balance())

```

part10:Nested Report Generator Create an outer function named generate_department_report().

Inside it, define nested functions to calculate:

Number of employees Total salary Average salary Average performance score Return the complete department report as a dictionary.

```python
def generate_department_report(department):
    dept_employees = list(filter(lambda emp: emp["department"] == department, employees))
    def employee_count():
        return len(dept_employees)
    def total_salary():
        return sum(emp["salary"] for emp in dept_employees)

    def average_salary():
        if len(dept_employees) == 0:
            return 0
        return total_salary()/employee_count()
    def average_performance():
        if len(dept_employees) == 0:
            return 0
        return sum(emp["performance_score"] for emp in dept_employees) / employee_count()
    return {
        "Department": department,
        "Number of Employees": employee_count(),
        "Total Salary": total_salary(),
        "Average Salary": average_salary(),
        "Average Performance Score": average_performance()
    }

report = generate_department_report("IT")
print("Department Report")
for key, value in report.items():
    print(key, ":", value)
```

```python
#1. Using map() and a lambda expression, calculate the square of every number.
n=[1,2,3,4,5,6]
r=map(lambda x:x**2,n)
print(list(r))
```

```python
#2. Using map() and a lambda expression, calculate the cube of every number.
n1=[2,3,4,5,6]
r1=map(lambda r1:r1**3,n1)
print(list(r1))
```

```python
#3. Using map(), add 5 to every number.
def even(x):
  return x+5
n2=[10,20,30,40,50]
r2=map(even,n2)
print(list(r2))

```

```python
#4. Using map(), subtract 10 from every number.
def sub(n3):
  return n3-10
n3=[15,25,35,45,55]
r3=list(map(sub,n3))
print(r3)
```

```python
#5. Using map(), multiply every number by 100.
def mul(n4):
  return n4*100
n4=[1,5,10,15,20]
r4=list(map(mul,n4))
print(r4)
```

```python
#6. Using map(), convert every name to uppercase.
def up(n5):
  return n5.upper()
n5=["ravi","anil","meena","kiran","sneha"]
r5=list(map(up,n5))
print(r5)
```

```python
#7. Using map(), convert every name to lowercase.
def low(n6):
  return n6.lower()
n6= ["RAHUL", "DIVYA", "ARJUN", "AJAY"]
r6=list(map(low,n6))
print(r6)
```

```python
#8. Using map(), convert every name to title case.
def t(n7):
  return n7.title()
n7=["rahul kumar", "anil sharma", "meena reddy", "divya singh"]
r7=list(map(t,n7))
print(r7)
```

```python
#9. Using map(), calculate the length of every word.
def l(n8):
  return len(n8)
n8=["Python", "Azure", "Databricks", "Cloud", "AI"]
r8=list(map(l,n8))
print(r8)
```

```python
#10. Using map(), extract the first character of every word.
def wor(n9):
  return n9[0]
n9= ["Python", "Azure", "Docker", "Linux", "Spark"]
r9=list(map(wor,n9))
print(r9)
```

```python
#11. Using map(), extract the last character of every word.
def w(n10):
  return n10[-1]
n10=["Python", "Azure", "Docker", "Linux", "Spark"]
r10=list(map(w,n10))
print(r10)
```

```python
#12. Using map(), convert every Celsius temperature into Fahrenheit.

def c(n11):
  return (9/5)*n11+32
n11=[0, 10, 20, 30, 40]
r11=list(map(c,n11))
print(r11)
```

```python
#13. Using map(), calculate 18% GST for every price.
def p(n12):
  return n12*0.18
n12= [100, 250, 500, 1000, 2500]
r12=list(map(p,n12))
print(r12)
```

```python
#14. Using map(), calculate the final price after adding 18% GST.
def gst(n13):
  return (n13+(n13*0.18))
n13= [100, 250, 500, 1000, 2500]
r13=list(map(gst,n13))
print(r13)
```

```python
#15. Using map(), round every number to two decimal places.
def de(n14):
  return round(n14,2)
n14= [12.4567, 45.8976, 78.1234, 99.9999]
r14=list(map(de,n14))
print(r14)
```

```python
#16. Using map(), convert every number into its absolute value.
def ad(n15):
  return abs(n15)
n15=[-10, 20, -35, 40, -55, 60]
r15=list(map(ad,n15))
print(r15)
```

```python
#17. Using map(), convert every integer into a string.
def a(n16):
  return str(n16)
n16=[10, 20, 30, 40, 50]
r16=list(map(a,n16))
print(r16)
```

```python
#18. Using map(), remove leading and trailing spaces from every name.
def sp(n17):
  return n17.strip()
n17=["  Ravi  ", " Anil", "Meena ", "  Kiran "]
r17=list(map(sp,n17))
print(r17)
```

```python
#19. Using map(), extract only employee names.
def emp(n18):
  return n18["name"]
n18= [
    {"id": 101, "name": "Rahul", "salary": 45000},
    {"id": 102, "name": "Anil", "salary": 65000},
    {"id": 103, "name": "Sneha", "salary": 72000},
    {"id": 104, "name": "Kiran", "salary": 82000}
]
r18=list(map(emp,n18))
print(r18)
```

```python
#20. Using map(), create a new list of dictionaries containing:

#Employee name
#Original salary
#Salary after a 10% increment
#Do not modify the original list.

def emp1(n19):
  return {
      "name":n19["name"],
      "salary":n19["salary"],
      "salary_inc":n19["salary"]*1.10,

  }

n19=[
    {"id": 101, "name": "Rahul", "salary": 45000},
    {"id": 102, "name": "Anil", "salary": 65000},
    {"id": 103, "name": "Sneha", "salary": 72000},
    {"id": 104, "name": "Kiran", "salary": 82000}
]
r19=list(map(emp1,n19))
print(r19)




```

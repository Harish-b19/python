```python
#1. Using filter(), extract all even numbers.
def even(n):
  return n%2==0

n= [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]
r=list(filter(even,n))
print(r)
```

```python
#2. Using filter(), extract all odd numbers.

def odd(n1):
  return n1%2
n1=[12, 45, 67, 89, 23, 56, 91, 34, 78, 100]
r=list(filter(odd,n1))
print(r)
```

```python
#3. Using filter(), extract all numbers greater than 50.

def great(n2):
  return n2>50
n2=[12, 45, 67, 89, 23, 56, 91, 34, 78, 100]
r1=list(filter(great,n2))
print(r1)
```

```python
#4. Using filter(), extract all numbers less than 100.
def less(n3):
  return n3<100
n3=[120, 45, 167, 89, 230, 56, 91, 34, 178, 100]
r2=list(filter(less,n3))
print(r2)
```

```python
#5. Using filter(), extract all positive numbers.
def pos(n4):
  return n4>0
n4= [-15, -8, 0, 4, 12, -25, 37, 42]
r3=list(filter(pos,n4))
print(r3)
```

```python
#6. Using filter(), extract all negative numbers.
def neg(n5):
  return n5<0
n5= [-15, -8, 0, 4, 12, -25, 37, 42]
r4=list(filter(neg,n5))
print(r4)
```

```python
#7. Using filter(), remove all zero values
def z(n6):
  return n6==0
n6=[ -15, -8, 0, 4, 12, 0, -25, 37, 0]
r5=list(filter(z,n6))
print(r5)
```

```python
#8. Using filter(), extract words containing more than five characters
def ext(n7):
  return len(n7)>5
n7=["python", "azure", "ai", "developer", "cloud", "analytics"]
r6=list(filter(ext,n7))
print(r6)
```

```python
#9. Using filter(), extract names beginning with "A".
def st(n8):
  return n8.startswith("A")
n8= ["Anil", "Rahul", "Ajay", "Meena", "Arjun", "Kiran", "Akhil"]
r7=list(filter(st,n8))
print(r7)
```

```python
#10. Using filter(), extract names ending with "n".
def end(n9):
  return n9.endswith("n")
n9= ["Kiran", "Mohan", "Arjun", "Rajan", "Meena", "Varun"]
r8=list(filter(end,n9))
print(r8)
```

```python
#11. Using filter(), extract all multiples of 5.
def extm(n10):
  return n10%5==0
n10=[10, 15, 18, 20, 25, 30, 32, 40, 45, 50]
r9=list(filter(extm,n10))
print(r9)
```

```python
#12. Using filter(), extract numbers divisible by both 3 and 5.
def extd(n11):
  return ((n11%3==0) and (n11%5==0))
n11= [3, 5, 10, 15, 18, 30, 42, 45, 60, 75]
r10=list(filter(extd,n11))
print(r10)
```

```python
#13. Using filter(), extract all palindrome words.
def pal(n12):
  return n12==n12[::-1]
n12=["madam", "python", "level", "azure", "radar", "cloud", "civic"]
r11=list(filter(pal,n12))
print(r11)
```

```python
#14. Using filter(), extract email addresses ending with ".com".
def e(n13):
  return n13.endswith(".com")
n13= [
    "john@gmail.com",
    "rahul@yahoo.com",
    "admin@python.com",
    "support@azure.org",
    "sales@company.com"
]
r12=list(filter(e,n13))
print(r12)
```

```python
#15. Using filter(), extract all CSV files.
def exts(n14):
  return n14.endswith("csv")
n14= [
    "sales.csv",
    "employees.xlsx",
    "products.csv",
    "report.pdf",
    "customers.csv",
    "notes.txt"
]
r13=list(filter(exts,n14))
print(r13)
```

```python
#16. Using filter(), extract employees whose salary is greater than 60000.
def sal(n15):
  return (n15["salary"]>60000)
n15=  [
    {"name": "Rahul", "salary": 45000},
    {"name": "Anil", "salary": 65000},
    {"name": "Sneha", "salary": 72000},
    {"name": "Kiran", "salary": 82000},
    {"name": "Meena", "salary": 55000}
]
r14=list(filter(sal,n15))
print(r14)
```

```python
#18. Using filter(), extract products that are currently in stock.
def stk(n17):
  return (n17["stock"]>0)
n18= [
    {"name": "Laptop", "price": 55000, "stock": 5},
    {"name": "Mouse", "price": 600, "stock": 25},
    {"name": "Keyboard", "price": 1500, "stock": 0},
    {"name": "Monitor", "price": 18000, "stock": 8},
    {"name": "Printer", "price": 12000, "stock": 0}
]
r17=list(filter(stk,n18))
print(r17)
```

```python
#19. Using filter(), extract people who are eligible to vote.
def vote(n18):
  return n18["age"]>18
n18= [
    {"name": "Rahul", "age": 17},
    {"name": "Anil", "age": 25},
    {"name": "Sneha", "age": 16},
    {"name": "Kiran", "age": 32},
    {"name": "Meena", "age": 18}
]
r17=list(filter(vote,n18))
print(r17)
```
